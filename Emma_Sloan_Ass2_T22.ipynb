```python
import pandas as pd

def get_config_metrics(results, config_label):
    class_names = results.names  # e.g. {0: 'Cyclist', 1: 'Pedestrian', 2: 'Car'}
    row = {
        'Config': config_label,
        'mAP50': results.box.map50,
        'mAP50-95': results.box.map,
    }
    for i, name in class_names.items():
        row[f'{name}_AP50'] = results.box.ap50[i]
        row[f'{name}_AP50-95'] = results.box.ap[i]
    return row
```

# Training 4 configurations

Config 1 - Hyperparameter settings provided in brief

```python
model_1 = YOLO("yolo11n.pt")

train_results_1 = model_1.train(
    data=dataset_yaml,
    epochs=20,
    imgsz=640,
    batch=16,
    lr0=0.001,
    freeze=10,
    workers=2,
    cache="ram",
    device=0,
    exist_ok=True,
    pretrained=True,
)
```

```python
# get metrics for validation set

val_metrics_1 = model_1.val(
    data=dataset_yaml,
    split="val",
    imgsz=640,
    conf=0.001,
    iou=0.7,
    exist_ok=True)
```

Config 2 - More Epochs

```python
model_2 = YOLO("yolo11n.pt")

train_results_2 = model_2.train(
    data=dataset_yaml,
    epochs=40, # more epochs 40
    imgsz=640,
    batch=16,
    lr0=0.001,
    freeze=10,
    workers=2,
    cache="ram",
    device=0,
    exist_ok=True,
    pretrained=True,
)
```

```python
# get metrics for validation set

val_metrics_2 = model_2.val(
    data=dataset_yaml,
    split="val",
    imgsz=640,
    conf=0.001,
    iou=0.7,
    exist_ok=True)
```

Config 3 - Reduced Freeze

```python
model_3 = YOLO("yolo11n.pt")

train_results_3 = model_3.train(
    data=dataset_yaml,
    epochs=20,
    imgsz=640,
    batch=16,
    lr0=0.001,
    freeze=5, # reduced freeze from 10 to 5 layers
    workers=2,
    cache="ram",
    device=0,
    exist_ok=True,
    pretrained=True,
)
```

Config 4 - combined more epochs and reduced freeze

```python
model_4 = YOLO("yolo11n.pt")

train_results_4 = model_4.train(
    data=dataset_yaml,
    epochs=40, # increased epochs from 20 to 40
    imgsz=640,
    batch=16,
    lr0=0.001,
    freeze=5, # reduced freeze from 10 to 5 layers
    workers=2,
    cache="ram",
    device=0,
    exist_ok=True,
    pretrained=True,
)
```

# Evaluate the best configuration

```python
rows = []
rows.append(get_config_metrics(val_metrics_1, 'Config 1 (baseline)'))
rows.append(get_config_metrics(val_metrics_2, 'Config 2 (more epochs)'))
rows.append(get_config_metrics(val_metrics_3, 'Config 3 (reduced freeze)'))
rows.append(get_config_metrics(val_metrics_4, 'Config 4 (combined)'))

comparison_table = pd.DataFrame(rows)
comparison_table
```
